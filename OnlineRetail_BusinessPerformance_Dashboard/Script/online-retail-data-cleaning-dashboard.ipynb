```python
import pandas as pd

# 1. Load the raw dataset (.xlsx format)
# If the file has both years across multiple sheets, we load and combine them:
excel_file = 'online_retail_II.xlsx'
sheets_dict = pd.read_excel(excel_file, sheet_name=None)  # loads all sheets into a dictionary
raw_df = pd.concat(sheets_dict.values(), ignore_index=True)

# 2. Data Cleaning Pipeline
# Remove transactions with missing Customer IDs
df_clean = raw_df.dropna(subset=['Customer ID']).copy()

# Exclude cancelled orders (invoices starting with 'C')
df_clean = df_clean[~df_clean['Invoice'].astype(str).str.startswith('C')]

# Keep only positive quantities and unit prices
df_clean = df_clean[(df_clean['Quantity'] > 0) & (df_clean['Price'] > 0)]

# 3. Data Transformation & Feature Engineering
# Cast Customer ID to integer
df_clean['Customer ID'] = df_clean['Customer ID'].astype(int)

# Standardize price precision and compute total revenue
df_clean['Price'] = df_clean['Price'].round(2)
df_clean['Revenue'] = (df_clean['Quantity'] * df_clean['Price']).round(2)

# 4. Export Cleaned Data
# Explicitly enforce 2-decimal floating point format to prevent precision artifacts
output_filename = 'online_retail_cleaned.csv'
df_clean.to_csv(output_filename, index=False, float_format='%.2f')

print(f"Pipeline executed successfully. Processed rows: {len(df_clean):,} saved to '{output_filename}'.")
```
